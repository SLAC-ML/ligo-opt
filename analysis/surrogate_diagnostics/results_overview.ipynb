# Surrogate diagnostics: results overview

The saved results are split into two notebooks to keep each report smaller:

- [Run 2 results](results_overview_run2.ipynb): data and feature audits, local fits,
  checkpoint evaluation, and the generalization workflow check.
- [Run 3 results](results_overview_run3.ipynb): data and feature audits, comparisons
  against run 2 checkpoints, the original PSO trajectory, fitting, and transfer results.

Both reports include saved plots and tables. Run all cells with the **Python (ligoopt)**
kernel to refresh them from `analysis/surrogate_diagnostics/outputs/`. Each report
includes output folders for its run automatically and has its own setup and filters.
Refreshing reads saved artifacts; it does not train models or run Finesse simulations.
